```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
train = pd.read_csv("/Users/srivarshinib/data/raw/train.csv")
test = pd.read_csv("/Users/srivarshinib/data/raw/test.csv")

print("Train shape:", train.shape)
print("Test shape:", test.shape)
```

```python
# Quick look at columns & types
print(train.info())
```

```python
# Count missing values
missing = train.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print("Missing values:\n", missing)

# Visualize top 20 missing
plt.figure(figsize=(10,6))
sns.barplot(x=missing[:20].values, y=missing[:20].index)
plt.title("Top Missing Values in Train Dataset")
plt.show()
```

```python
# Summary statistics
print(train["SalePrice"].describe())

# Histogram
plt.figure(figsize=(8,5))
sns.histplot(train["SalePrice"], bins=40, kde=True)
plt.title("Distribution of Sale Price")
plt.show()

# Boxplot for outliers
plt.figure(figsize=(8,5))
sns.boxplot(x=train["SalePrice"])
plt.title("Sale Price - Outlier Detection")
plt.show()
```
